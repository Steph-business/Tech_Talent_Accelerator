## 🌟 Exercice 2 : Architecture LLM et scénarios d'application
Pour chacun, décrivez (a) les différences architecturales fondamentales, (b) une application réelle, (c) pourquoi elle convient.

### BERT
- **Architecture** : Basé sur l'encodeur du Transformeur, il utilise une attention bidirectionnelle pour lire le texte de gauche à droite et de droite à gauche simultanément. Pré-entraîné via le Masked Language Modeling (MLM).
- **Application** : Recherche Google (Google Search).
- **Pourquoi** : Sa nature bidirectionnelle lui permet de comprendre parfaitement le contexte global d'une requête de recherche, ce qui est crucial pour lever les ambiguïtés sémantiques.

### GPT
- **Architecture** : Basé sur le décodeur du Transformeur, il utilise une attention causale (masquée) pour prédire le jeton suivant de manière autorégressive.
- **Application** : Assistants de rédaction créative ou chatbots (ex: ChatGPT).
- **Pourquoi** : Conçu spécifiquement pour la génération de texte fluide et cohérente, il excelle dans la poursuite logique d'un récit ou d'un dialogue.

### T5
- **Architecture** : Architecture complète Encodeur-Décodeur qui traite chaque tâche PNL comme un problème de conversion de "texte-à-texte".
- **Application** : Résumé automatique de longs articles.
- **Pourquoi** : L'encodeur comprend l'intégralité du texte source tandis que le décodeur génère un nouveau texte condensé, ce qui est idéal pour la réécriture et la synthèse.

## 🌟 Exercice 3 : Avantages et éthique du pré-entraînement
- **Avantages** :
  1) **Généralisation améliorée** : Le modèle apprend des structures linguistiques universelles sur de vastes corpus, ce qui lui permet de mieux réagir à des phrases inédites.
  2) **Moins de données étiquetées** : Puisque le modèle connaît déjà la langue, il n'a besoin que de quelques exemples spécifiques (Few-shot) pour apprendre une nouvelle tâche.
  3) **Affinage plus rapide** : On ne repart pas de zéro ; l'ajustement des poids pour une tâche précise est beaucoup plus court que l'entraînement initial.
  4) **Apprentissage par transfert** : Les connaissances acquises sur une tâche (ex: prédire des mots manquants) sont transférées à d'autres (ex: analyse de sentiments).
  5) **Robustesse** : L'exposition à une grande diversité de styles et de sujets rend le modèle plus stable face aux variations de langage.

- **Préoccupations éthiques** : Les modèles peuvent mémoriser des données privées, amplifier des stéréotypes sexistes ou raciaux présents dans les données d'entraînement, ou être utilisés pour générer des "Deepfakes" textuels convaincants.
- **Atténuations** : Utiliser le RLHF (Reinforcement Learning from Human Feedback) pour aligner le modèle sur des valeurs humaines, filtrer les jeux de données en amont, et implémenter des garde-fous (red teaming) pour bloquer les sorties toxiques.

## 🌟 Exercice 4 : Plongée profonde dans l'architecture des transformeurs
### Auto-attention et attention multi-têtes
- **Flux Q/K/V** : Chaque mot est projeté en trois vecteurs : Query (requête), Key (clé) et Value (valeur). Le produit scalaire entre Q et K détermine un score d'affinité, normalisé par Softmax pour obtenir des poids d'attention. Ces poids sont ensuite multipliés par V pour obtenir la représentation pondérée du mot dans son contexte.
- **Attention multi-têtes** : L'utilisation de plusieurs têtes permet au modèle de prêter attention à différents types de relations simultanément (syntaxiques, sémantiques, temporelles) dans différents sous-espaces de représentation.
- **Exemple de phrase** : "L'animal n'a pas traversé la rue car il était trop fatigué."
  - Tête 1 : Pourrait lier "il" à "animal" (résolution de coréférence).
  - Tête 2 : Pourrait lier "fatigué" à "animal" (attribution d'état).

### Objectifs de pré-entraînement
- **MLM vs CLM** : Le Masked Language Modeling (MLM) cache des mots au milieu de la phrase (bidirectionnel, ex: BERT), tandis que le Causal Language Modeling (CLM) prédit le mot suivant (unidirectionnel, ex: GPT).
- **Quand préférer MLM vs CLM** : On préfère le MLM pour la compréhension fine et l'extraction d'information (classification), et le CLM pour la génération de texte fluide.
- **NSP (Next Sentence Prediction)** : Utilisé initialement pour aider BERT à comprendre les relations entre phrases, il a été abandonné par beaucoup car jugé trop simple ou moins efficace que l'entraînement sur de plus longues séquences (comme dans RoBERTa).

### Sélection du modèle de transformeur
- **Sentiment sur les critiques** : Encodeur-seulement (type BERT). Justification : Excellent pour classer et extraire le sens global d'un texte fixe.
- **Chatbot conversationnel** : Décodeur-seulement (type GPT). Justification : Optimisé pour la génération fluide et la cohérence du dialogue.
- **Traduction technique** : Encodeur-Décodeur (type T5/BART). Justification : Nécessite une compréhension complète de la source avant de générer une cible structurée.

### Encodage positionnel
- **But** : Les Transformeurs traitent tous les mots en parallèle et n'ont pas de notion intrinsèque d'ordre. L'encodage positionnel ajoute un signal mathématique pour indiquer la position relative ou absolue des mots.
- **Échec exemplaire** : Sans positions, les phrases "Le chat mange la souris" et "La souris mange le chat" auraient exactement la même représentation vectorielle.

## 🌟 Exercice 6 : Température Softmax : Le régulateur d'aléatoire
### 1) Scénarios de température
- **T=0.2** : Très conservateur. Le modèle choisit presque toujours le mot le plus probable. Utile pour la précision factuelle.
- **T=1.5** : Très créatif, voire chaotique. Augmente la probabilité des mots rares, ce qui peut mener à des hallucinations.
- **T=1.0** : Température par défaut. Équilibre entre probabilité apprise et diversité.

### 2) Conception d'application
- **Histoires pour enfants** : Température moyenne-haute (0.7 - 0.9). On veut de l'originalité et de la magie tout en restant grammaticalement cohérent.
- **Résumés financiers** : Température très basse (0.1 - 0.2). On ne veut aucune invention ; le modèle doit rester strictement fidèle aux chiffres fournis.

### 3) Température et biais
- La température peut amplifier les biais : à T basse, le modèle se replie sur les associations les plus fréquentes vues durant l'entraînement (souvent des stéréotypes).
- *Exemple* : Pour le prompt "Le médecin est...", une température basse pourrait systématiquement compléter par "un homme", alors qu'une température plus élevée permettrait d'explorer d'autres possibilités, bien que cela n'élimine pas le biais fondamental du modèle.

### (Optionnel) Démo de génération rapide à différentes températures
Note : Nécessite `transformers` et le téléchargement du modèle ; passez si hors ligne.

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch

model_name = "gpt2"  # TODO: choisissez un LM causal
prompt = "L'intelligence artificielle sera"  # TODO: écrivez votre propre prompt

pipe = pipeline(
    "text-generation",
    model=AutoModelForCausalLM.from_pretrained(model_name),
    tokenizer=AutoTokenizer.from_pretrained(model_name),
    device=0 if torch.cuda.is_available() else -1,
)

for T in [0.2, 1.0, 1.5]:
    out = pipe(prompt, max_new_tokens=40, temperature=T, do_sample=True)
    print("--- Température :", T)
    print(out[0]["generated_text"])  # Observez comment le style change avec T
```
