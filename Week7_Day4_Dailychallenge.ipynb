### 1. Compréhension de l'évaluation des LLM

**Pourquoi l'évaluation des LLM est-elle complexe ?**
Contrairement aux logiciels traditionnels qui ont des sorties déterministes (une entrée = une sortie fixe), les LLM sont stochastiques (probabilistes). La même question peut donner des réponses différentes. De plus, la 'qualité' du langage est subjective.

**Sécurité des LLM :**
Il est crucial d'évaluer la sécurité pour éviter la génération de contenus toxiques, de biais discriminatoires ou de conseils dangereux.

**Tests adverses (Adversarial Testing) :**
Cela consiste à essayer de 'tromper' le modèle avec des entrées piégées pour identifier ses faiblesses et renforcer sa robustesse.

**Limites des métriques automatisées :**
Les métriques comme BLEU ou ROUGE se basent sur la similarité des mots et non sur le sens profond. L'évaluation humaine reste la référence pour juger la pertinence et la fluidité.

```python
# Installation de la bibliothèque manquante pour ROUGE
!pip install rouge-score

import nltk
from nltk.translate.bleu_score import sentence_bleu
from rouge_score import rouge_scorer

# Téléchargement des outils nécessaires pour la tokenisation
nltk.download('punkt')

# --- Exercice 2: Calcul de BLEU ---
# Référence (la vérité terrain)
reference = ["Despite the increasing reliance on artificial intelligence in various industries, human oversight remains essential to ensure ethical and effective implementation.".split()]
# Génération du modèle
generated = "Although AI is being used more in industries, human supervision is still necessary for ethical and effective application.".split()

# Calcul du score BLEU (mesure la précision des n-grammes)
bleu_score = sentence_bleu(reference, generated)
print(f"Score BLEU : {bleu_score:.4f}")

# --- Exercice 2: Calcul de ROUGE ---
# Création d'un calculateur pour ROUGE-L (basé sur la plus longue sous-séquence commune)
scorer = rouge_scorer.RougeScorer(['rougeL'], use_stemmer=True)
ref_text = "In the face of rapid climate change, global initiatives must focus on reducing carbon emissions and developing sustainable energy sources to mitigate environmental impact."
gen_text = "To counteract climate change, worldwide efforts should aim to lower carbon emissions and enhance renewable energy development."

# Calcul du score
scores = scorer.score(ref_text, gen_text)
print(f"Score ROUGE-L : {scores['rougeL'].fmeasure:.4f}")
```

### 3. Analyse de la Perplexité

**Comparaison Modèle A et B :**
La perplexité est inversement proportionnelle à la probabilité.
- Formule simplifiée : $PPL(W) = \frac{1}{P(word)}$
- Modèle A : $1 / 0.8 = 1.25$
- Modèle B : $1 / 0.4 = 2.5$

**Résultat :** Le **Modèle A** a la perplexité la plus basse. Une perplexité plus faible signifie que le modèle est moins 'perplexe' (plus confiant) et prédit mieux la suite du texte.

**Note sur le score de 100 :** Une perplexité de 100 signifie que le modèle hésite comme s'il devait choisir entre 100 mots équiprobables. Pour l'améliorer, on peut augmenter la taille du dataset d'entraînement ou affiner l'architecture du modèle.

### 6. Analyse Comparative des Méthodes d'Évaluation

**Tâche choisie :** Traduction Automatique (Machine Translation)

| Métrique | Avantages | Inconvénients |
| :--- | :--- | :--- |
| **BLEU** | Rapide, peu coûteux, standard de l'industrie. | Ne comprend pas le sens, pénalise les synonymes. |
| **BERTScore** | Capture la similarité sémantique (le sens) via des embeddings. | Plus lent car nécessite un modèle de Deep Learning. |
| **Évaluation Humaine** | La plus précise pour la fluidité et les nuances culturelles. | Très lente, coûteuse et subjective. |

**Quelle métrique est la plus appropriée ?**
Pour la traduction automatique, le **BERTScore** est souvent préférable aujourd'hui car il reconnaît que deux phrases peuvent avoir le même sens avec des mots différents. Cependant, l'**Évaluation Humaine** reste indispensable pour la validation finale d'un produit destiné aux utilisateurs.
