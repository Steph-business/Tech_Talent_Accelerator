# Défi quotidien : Incitations précises pour un contrôle et une qualité de production optimaux

Ce notebook contient les étapes du défi d'ingénierie de prompt pour la mise à jour de la politique de sécurité de l'accès à distance.

```python
input_text = """Les employés doivent s'assurer que tout accès distant aux systèmes internes est établi via le VPN sécurisé approuvé. En aucun cas, des connexions non sécurisées ou des appareils personnels dépourvus de protection des terminaux ne doivent être utilisés pour accéder à des données confidentielles ou à des communications sensibles."""
print("Texte d'entrée chargé.")
```

## Étape 1 : Créer une invite de contrôle

Voici le prompt conçu pour réécrire le texte avec un ton amical, des puces, et une contrainte de longueur (< 75 mots).

```python
prompt_step_1 = f"""
Agis comme un expert en communication interne. Réécris le texte suivant en respectant ces consignes :
1. Utilise un ton amical et clair.
2. Organise les informations sous forme de liste à puces.
3. Paraphrase le contenu sans utiliser de citations directes.
4. Le texte final doit faire moins de 75 mots.

Texte source : {input_text}
"""

print("Prompt Étape 1 :")
print(prompt_step_1)
```

## Étape 4 : Analyse approfondie (Public Juniors)

Prompt pour transformer le texte pour des stagiaires juniors.

```python
prompt_juniors = f"""
Réécris ce paragraphe pour des stagiaires juniors :
- Utilise un langage simple et des phrases courtes.
- Maximum 4 puces.
- Évite le jargon d'entreprise ou juridique.
- Ton encourageant et informatif.

Texte source : {input_text}
"""

print("Prompt pour Juniors :")
print(prompt_juniors)
```

## Étape 5 : Variante d'extraction de citation

Ce prompt vise à extraire la phrase la plus essentielle du texte original.

```python
prompt_citation = f"""
Extrait une seule citation directe du texte suivant qui résume le mieux la politique de sécurité essentielle :

{input_text}
"""

print("Prompt d'extraction de citation :")
print(prompt_citation)
```

### Analyse des Citations vs Paraphrase

**Dans quel type de communication interne la citation serait-elle plus appropriée ?**
Une citation directe est idéale pour les documents officiels, les chartes de conformité signées ou les affiches de rappel réglementaire où l'exactitude juridique est primordiale.

**Dans quelles circonstances les citations peuvent-elles présenter un risque ?**
Les citations peuvent être risquées si elles sont trop denses ou pleines de jargon, car elles risquent d'être ignorées par le personnel non technique, ou si elles sont sorties de leur contexte global, perdant ainsi les nuances de la politique complète.
