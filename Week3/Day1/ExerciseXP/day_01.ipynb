```python
# Exercise 2 :Normalize a 3x3 random matrix (subtract the mean and divide by the standard deviation of the matrix).
array_random = np.random.rand(3, 3)
mean = np.mean(array_random)
std = np.std(array_random)
normalized_array = (array_random - mean) / std if std != 0 else array_random - mean
normalized_array
```
