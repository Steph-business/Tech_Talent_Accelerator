```python
import numpy as np         
import pandas as pd         
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
temperature = np.random.uniform(-5, 35, size=(10,12))
temperature
```

```python
cities = ["Abidjan", "Yamoussoukro", "Bouaké", "Daloa", "San-Pédro",
          "Korhogo", "Man", "Gagnoa", "Divo", "Abengourou"]

mouths = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
        "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

df = pd.DataFrame(temperature, index=cities, columns=mouths)
df
```

```python
annual_mean = df.mean(axis=1)
annual_mean
```

```python
hottest_city = annual_mean.idxmax()
coldest_city = annual_mean.idxmin()

print(f"The hottest city is {hottest_city} with an annual mean temperature of {annual_mean[hottest_city]:.2f}°C.")
print(f"The coldest city is {coldest_city} with an annual mean temperature of {annual_mean[coldest_city]:.2f}°C.")
```

```python
plt.figure(figsize=(10, 6))
plt.bar(annual_mean.index, annual_mean.values, color='orange')
plt.title('Annual Mean Temperature of Ivorian Cities')
plt.xlabel('Cities')
plt.ylabel('Annual Mean Temperature (°C)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

for city in df.index:
    plt.plot(df.columns, df.loc[city], marker='o', label=city)

plt.title('Monthly Temperatures of Ivorian Cities')
plt.xlabel('Months')
plt.ylabel('Temperature (°C)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')   
plt.tight_layout()

plt.show()
```

```python

# Temperature Trend Analysis Across Ivorian Cities


# 1. OBJECTIVE
# Analyze and visualize average monthly temperature trends across several
# cities, combining NumPy, Pandas, and Matplotlib in a single workflow.
#
# 2. METHODOLOGY
# - Data generation: a synthetic dataset was created with NumPy using
#   np.random.uniform(-5, 35, size=(10, 12)), producing average monthly
#   temperatures (in °C) for 10 Ivorian cities over 12 months.
#   A fixed seed (np.random.seed(42)) was used for reproducibility.
# - Data structuring: the NumPy array was converted into a Pandas DataFrame,
#   with city names as the index (rows) and months as the columns.
# - Analysis: the annual mean temperature of each city was computed with
#   df.mean(axis=1). The hottest and coldest cities were identified with
#   idxmax() and idxmin().
# - Visualization: a bar chart compares the annual mean temperature per
#   city, and a line chart shows the monthly temperature trend of each city.
#
# 3. KEY FINDINGS
# - Hottest city: Gagnoa, with an annual mean temperature of 17.61°C.
# - Coldest city: Yamoussoukro, with an annual mean temperature of 9.57°C.
# - The gap between the warmest and coldest city is about 8.03°C.
#
# 4. OBSERVED TRENDS
# - Because the temperatures are generated uniformly at random and
#   independently for each month, the monthly curves fluctuate sharply and
#   show no real seasonal pattern (no smooth summer/winter cycle).
# - All annual means fall within a narrow band (~9-18°C). This is expected:
#   averaging 12 uniform random values tends toward the middle of the range.
# - A real dataset would instead show a gradual, smooth seasonal trend.
#
# 5. CONCLUSION
# This project demonstrates a complete data-analysis workflow - generate,
# structure, analyze, and visualize - by combining NumPy, Pandas, and
# Matplotlib. Although the data is synthetic, the same techniques apply
# directly to real-world weather datasets.
```
