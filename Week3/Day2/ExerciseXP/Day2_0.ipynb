## Exercise 1 : Matrix Operations (determinant & inverse)

```python
# Build a 3x3 matrix with non-zero determinant
A = np.array([
    [3, 1, 2],
    [4, 5, 6],
    [7, 8, 10]
])

det = np.linalg.det(A)
print(f'Matrix A:\n{A}\n')
print(f'Determinant of A: {det:.4f}')

if not np.isclose(det, 0):
    A_inv = np.linalg.inv(A)
    print(f'\nInverse of A:\n{A_inv}')
    # Sanity check: A @ A_inv should be ~ identity
    print(f'\nA @ A_inv (should be identity):\n{np.round(A @ A_inv, 6)}')
else:
    print('A is singular -> no inverse.')
```

## Exercise 5 : Image Representation

A grayscale image is simply a **2D NumPy array** whose values represent pixel intensities (usually 0 = black, 255 = white).  
A color (RGB) image is a **3D array** of shape `(height, width, 3)` — one channel per primary color.  
An RGBA image has a 4th channel for opacity.

```python
# Build a simple 5x5 grayscale image (uint8 in [0, 255])
gray_image = np.array([
    [  0,  50, 100, 150, 200],
    [ 50, 100, 150, 200, 255],
    [100, 150, 200, 255, 200],
    [150, 200, 255, 200, 150],
    [200, 255, 200, 150, 100]
], dtype=np.uint8)

print('Shape         :', gray_image.shape)
print('dtype         :', gray_image.dtype)
print('Min / Max     :', gray_image.min(), '/', gray_image.max())

plt.figure(figsize=(3, 3))
plt.imshow(gray_image, cmap='gray', vmin=0, vmax=255)
plt.title('5x5 grayscale image')
plt.axis('off')
plt.show()
```

## Exercise 6 : Basic Hypothesis Testing

```python
# Productivity before and after the training program
productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after  = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Hypotheses:
#   H0: the training program has NO effect       -> mean(after - before) == 0
#   H1: the training program improves productivity -> mean(after - before) > 0
differences = productivity_after - productivity_before

mean_diff = differences.mean()
std_diff  = differences.std(ddof=1)
n         = len(differences)

# Compute the t-statistic manually (paired-sample t-test)
t_stat = mean_diff / (std_diff / np.sqrt(n))
# Cross-check using SciPy
t_scipy, p_scipy = stats.ttest_rel(productivity_after, productivity_before)

print(f'Mean difference (after - before) : {mean_diff:.3f}')
print(f't (manual)                       : {t_stat:.4f}')
print(f't (scipy)                        : {t_scipy:.4f}')
print(f'p-value (two-sided, scipy)       : {p_scipy:.6f}')

alpha = 0.05
if p_scipy < alpha:
    print(f'\np < {alpha} -> reject H0: the training program has a significant effect.')
else:
    print(f'\np >= {alpha} -> fail to reject H0.')
```

## Exercise 1 : Eigenvalues, Eigenvectors and Matrix Normalization

```python
# 5x5 matrix with random values
M = np.random.rand(5, 5)
print('Matrix M:')
print(np.round(M, 3))

# Eigenvalues and eigenvectors
eigvals, eigvecs = np.linalg.eig(M)
print('\nEigenvalues:')
print(np.round(eigvals, 4))
print('\nEigenvectors (columns):')
print(np.round(eigvecs, 4))
```

```python
# Two common matrix normalization approaches:
# (a) Min-Max -> map every entry into [0, 1]
M_minmax = (M - M.min()) / (M.max() - M.min())

# (b) Z-score normalization (per column)
M_zscore = (M - M.mean(axis=0)) / M.std(axis=0)

print('Min-Max normalized:')
print(np.round(M_minmax, 3))
print('\nZ-score normalized (per column):')
print(np.round(M_zscore, 3))
```

## Exercise 2 : Statistical Distribution Analysis

```python
sample = np.random.normal(loc=0, scale=1, size=10000)

skew = stats.skew(sample)
kurt = stats.kurtosis(sample)   # Fisher's definition: 0 for a normal distribution

plt.figure(figsize=(9, 4))
plt.hist(sample, bins=50, density=True, color='lightblue', edgecolor='black')
plt.title(f'Normal(0, 1) — skewness = {skew:.3f}, kurtosis = {kurt:.3f}')
plt.show()

print('Skewness ~ 0  -> symmetric distribution')
print('Kurtosis ~ 0  -> tails are as heavy as a normal distribution ("mesokurtic")')
```

## Exercise 3 : Time Series Forecasting (linear regression)

```python
monthly_sales_data = np.random.randint(100, 500, size=12)
months = np.arange(1, 13)

# Fit a simple linear regression: sales = a*month + b
slope, intercept = np.polyfit(months, monthly_sales_data, deg=1)

# Forecast the next 6 months
future_months = np.arange(13, 19)
forecast = slope * future_months + intercept

plt.figure(figsize=(10, 4))
plt.plot(months, monthly_sales_data, marker='o', label='Observed sales')
plt.plot(future_months, forecast, marker='x', linestyle='--', color='red', label='Forecast')
plt.plot(np.r_[months, future_months], slope * np.r_[months, future_months] + intercept,
         color='black', alpha=0.5, label='Trend line')
plt.title(f'Monthly sales + 6-month forecast  (slope = {slope:.2f})')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.legend()
plt.show()

print('Forecast for months 13-18:', np.round(forecast, 1))
```

## Exercise 4 — Pandas Data Aggregation

```python
data = {
    'Product': ['Product A', 'Product B', 'Product C', 'Product A', 'Product B'],
    'Sales':   [200, 150, 300, 250, 180],
    'Month':   ['January', 'February', 'March', 'April', 'May']
}
df_sales = pd.DataFrame(data)
print('DataFrame:')
print(df_sales)

# Group by product
by_product = df_sales.groupby('Product')['Sales'].agg(['sum', 'mean', 'count']).round(2)
print('\nSales aggregated per product:')
print(by_product)

# Group by month
by_month = df_sales.groupby('Month')['Sales'].sum()
print('\nTotal sales per month:')
print(by_month)
```

## Exercise 2 : Conditional Array Operations

## Exercise 3 : Data Normalization

```python
x = np.random.randint(0, 100, size=15).astype(float)

# Manual z-score normalization
normalized_manual = (x - x.mean()) / x.std()

# Same thing with sklearn's StandardScaler
from sklearn.preprocessing import StandardScaler
normalized_sklearn = StandardScaler().fit_transform(x.reshape(-1, 1)).ravel()

print('Original         :', x)
print('Manual z-score   :', np.round(normalized_manual, 4))
print('StandardScaler   :', np.round(normalized_sklearn, 4))
print('\nMean of normalized data:', round(normalized_manual.mean(), 6))
print('Std  of normalized data:', round(normalized_manual.std(), 6))
```

## Exercise 4 : Pearson Correlation Coefficient

```python
x = np.random.rand(50)
y = 2 * x + np.random.normal(0, 0.1, size=50)   # strongly correlated with x

# NumPy returns the full correlation matrix
corr_matrix = np.corrcoef(x, y)
pearson_r = corr_matrix[0, 1]
print('Correlation matrix:')
print(np.round(corr_matrix, 4))
print(f'\nPearson r = {pearson_r:.4f}')

# Cross-check using SciPy (also returns the p-value)
r_sp, p_sp = stats.pearsonr(x, y)
print(f'SciPy r = {r_sp:.4f}  (p = {p_sp:.4e})')
```

## Exercise 5 : Time Series Trend Line

```python
# 15 days of random values (cumulative random walk for visual interest)
days   = np.arange(1, 16)
values = np.random.randn(15).cumsum() + 10

# Fit a linear trend (degree 1)
slope, intercept = np.polyfit(days, values, deg=1)
trend_line = slope * days + intercept

plt.figure(figsize=(10, 4))
plt.plot(days, values, marker='o', label='Time series', color='steelblue')
plt.plot(days, trend_line, color='red', linestyle='--',
         label=f'Trend  y = {slope:.2f} x + {intercept:.2f}')
plt.title('15-day time series with linear trend line')
plt.xlabel('Day')
plt.ylabel('Value')
plt.legend()
plt.show()
```
