### Exercice 2 : Choisir le bon modle de prompt (Classification)

```python
# Choix de la stratégie : Few-Shot Prompting
# Pourquoi ? Pour garantir la cohérence des catégories et le format de sortie.

few_shot_classifier = """
Tu es un agent de support. Catégorise le message client dans l'une de ces classes : [Facturation, Support Technique, Accès Compte, Autre].

Exemple 1 :
Message : 'Je n'arrive pas à me connecter.'
Classe : Accès Compte

Exemple 2 :
Message : 'Pourquoi ai-je été prélevé deux fois ?'
Classe : Facturation

Exemple 3 :
Message : [MESSAGE_CLIENT]
Classe :
"""
justification = "Le Few-Shot permet de guider le modèle sur le format exact attendu sans ambiguïté."
display(few_shot_classifier, justification) # Affiche le modèle et sa justification
```

### Exercice 3 : AlignedCoT pour rduire les hallucinations

```python
# Prompt AlignedCoT avec deux chemins de raisonnement distincts
aligned_cot_prompt = """
Problme : Calculer le cot total de 2 petits pots (2$), 3 moyens (4$) et 1 grand (6$).

Chemin 1 (Par type de pot) :
- Petits : 2 * 2$ = 4$
- Moyens : 3 * 4$ = 12$
- Grands : 1 * 6$ = 6$
- Somme : 4 + 12 + 6 = 22$

Chemin 2 (Groupement par addition successive) :
- 2 + 2 (petits) = 4
- 4 + 4 + 4 (moyens) = 12
- 6 (grand) = 6
- Total : 4 + 12 + 6 = 22$

Comparaison : Les deux mthodes arrivent au rsultat de 22$. Le rsultat est cohrent.
"""
print(aligned_cot_prompt) # Affiche le raisonnement align
```
